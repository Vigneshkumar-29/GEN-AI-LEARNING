```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough,RunnableParallel
from langchain_community.document_loaders import PyPDFLoader,TextLoader
import os
```

```python
parser = StrOutputParser()
```

```python
# os.environ['GEMINI_API_KEY'] = ''
```

```python
llm = ChatGoogleGenerativeAI(
    model='gemini-3.5-flash',
    api_key=os.getenv('GEMINI_API_KEY'),
)
```

## LCEL 1 - `Extracting the skills from the resume`

```python
skill_prompt = ChatPromptTemplate.from_messages([
    ("system","Extract the skills from the resume."),
    ("human","resume : {resume}")
    ]
)

skill_chain = skill_prompt | llm | parser #lcel
```

## LCEL 2 - `Summarize the about project/exp from the resume`

```python
summarize_prompt = ChatPromptTemplate.from_messages(
    [
    ("system","Summarize the experience from the resume."),
    ("human","resume : {resume}")
    ]
)
summarize_chain = summarize_prompt | llm | parser
```

## LCEL - `Find the missing skills from the resume for Data Science Role`

```python
missing_skills_prompt = ChatPromptTemplate.from_messages([
    ("system","For Data Science Role find the missing skills from the resume."),
    ("human","resume : {resume}")
])
missing_skills_chain = missing_skills_prompt | llm | parser
```

```python
ats = ChatPromptTemplate.from_messages(
    [
    ("system","You have to read the Resume and give the ATS score for the give Resume out of 100."),
    ("human","resume:{resume}")
    ]
)
ats_chain = ats | llm | parser
```

```python
final_chain = RunnableParallel(
    skills = skill_chain,
    summarize = summarize_chain,
    missing_skills = missing_skills_chain,
    ats_score = ats_chain
)
```
