```python
from sentence_transformers import SentenceTransformer
''' 
It uses encoder only transformer architecture and used for Embeddings.
It gives meaning full vectors.
'''
from langchain_text_splitters import RecursiveCharacterTextSplitter
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
data = open('data.txt').read()
data
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size = 100,
    chunk_overlap = 20
)
```

```python
chunks = splitter.split_text(data) #list of strings(chunks)
print(chunks,type(chunks),type(chunks[0]))
```

```python
embeddings = SentenceTransformer(
    model_name_or_path = 'sentence-transformers/all-miniLM-L6-V2'  #384
)
```

```python
embeddings.encode(chunks) #it returns list of vectors.
```
