```python
import contractions
import re
import string
import emoji
```

```python
data = open('data.txt',encoding='utf-8').read()
```

```python
# make the data into lower()

data = data.lower()
```

```python
# removing the extra spaces

data = re.sub(r'\s{2,}',' ',data)
```

```python
# contraction expanding the words

data = contractions.fix(data)
```

```python
# removing the special chars and symbols 
# removing punctuation

str_p = string.punctuation

str_p = str_p.replace('.','')

data = re.sub(f'[{str_p}]','',data)
```
