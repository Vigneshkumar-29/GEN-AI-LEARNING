```python
import re

with open ('unclean.txt','r',encoding='utf-8') as f:
    data = f.read()

# convert to lower case
    data = data.lower()

# removing the 2 or more space 
    data = re.sub(r'\s{2,}',' ',data)

# removing the spacial character , symbols , emoji , un-wanted character.
    data = re.sub(r'[^a-zA-Z0-9\s]','',data)          

# save the file
with open ('cleaned.txt','w',encoding='utf-8') as f:
    f.write(data)
```
