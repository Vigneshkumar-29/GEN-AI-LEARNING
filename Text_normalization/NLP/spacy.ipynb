```python
import spacy
nlp = spacy.load('en_core_web_sm')

tokens = nlp('''Apply the pipeline To some text.
                The Text can Span multiple senetences,
                and can contain arbitrary whitespace.
                Alignment into the original String is Preserved,.
                
                Running,cycling,trying ''')

print(type(tokens), tokens)
```

```python
for token in tokens:
    print(f'{token} => {token.lemma_}')
```

```python
for token in tokens:
    print(f'{token} => {token.is_stop}')
```

```python
new_tokens = [str (token) for token in tokens if not token.is_stop]         #compreshion 

new_tokens
```
