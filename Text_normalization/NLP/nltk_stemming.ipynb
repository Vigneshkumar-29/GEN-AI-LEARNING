```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
```

```python
words = [
    "playing",
    "played",
    "plays",
    "running",
    "runner",
    "easily",
    "fairies",
    "studies",
    "connected",
    "connection"
]

for word in words:
    print(f'{word} ----> {stemmer.stem(word)}')
```

```python
from nltk.stem import SnowballStemmer
Snowball_Stemmer = SnowballStemmer("english", ignore_stopwords=False)
```
