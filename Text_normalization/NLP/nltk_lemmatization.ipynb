```python
import nltk
from nltk.stem import WordNetLemmatizer
```

```python
tokens = ['houses','running','dogs','better','cats','flying','gees','studies','wolves','children']
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
words = 'sleeping'
t = lemmatizer.lemmatize(words,pos='v')
print(t)
```

```python
for word in tokens:
    print(word,lemmatizer.lemmatize(word,pos='v'))
```

```python
pos_tag_word = nltk.pos_tag(tokens)
pos_tag_word
#  we get the pos from using the pos_tag
#  from averaged_perceptron_tagger_eng
```

```python
def find_postag(tag):

    if tag.startswith('N'):
        return 'n'

    elif tag.startswith('V'):
            return 'v'

    elif tag.startswith('J'):
                return 'a'

    elif tag.startswith('R'):
                return 'r'

    else:
            return 'n'


```

```python
for word,tag in pos_tag_word:
    tagged_word = lemmatizer.lemmatize(word,pos=(find_postag(tag)))
    print(tagged_word)
```
