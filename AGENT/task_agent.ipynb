```python
from langchain_ollama import ChatOllama
from langchain_core.tools import tool
from langchain.agents import create_agent
import pandas as pd
```

```python
df = pd.DataFrame({
    'product':["Laptop", "Phone", "Tablet", "Laptop", "Phone"],
    "Sales": [50000, 30000, 20000, 45000, 35000]
})
```

```python
@tool
def get_total_sales() -> str:
    """Calculate the total sales."""

    total_sales = df["Sales"].sum()

    return f"Total Sales: {total_sales}"
```

```python
@tool
def get_highest_revenue() -> str:
    """Find the highest revenue transaction."""

    highest_revenue = df["Sales"].max()

    return f"Highest Revenue: {highest_revenue}"
```

```python
@tool
def get_best_product() -> str:
    """Find the best-selling product based on total sales."""

    sales_by_product = (
        df.groupby("Product")["Sales"]
        .sum()
        .sort_values(ascending=False)
    )

    best_product = sales_by_product.index[0]

    return f"Best-Selling Product: {best_product}"
```

```python
@tool
def get_business_highlights() -> str:
    """Generate important business highlights from the sales data."""

    total_sales = df["Sales"].sum()

    sales_by_product = (
        df.groupby("Product")["Sales"]
        .sum()
        .sort_values(ascending=False)
    )

    best_product = sales_by_product.index[0]

    highest_revenue = df["Sales"].max()

    return f"""
Business Highlights:

1. Total sales generated: {total_sales}

2. Best-selling product: {best_product}

3. Highest individual revenue transaction: {highest_revenue}

4. Business insight: {best_product} is the strongest-performing product and should receive attention for inventory, marketing, and promotional activities.
"""
```

```python
tools = [get_total_sales,get_highest_revenue,get_business_highlights,get_best_product]
```

```python
model = ChatOllama(
    model = 'qwen2.5:0.5b'
)
```

```python
agents = create_agent(model,tools)
```

```python
response = agents.invoke({
    "messages":[
        ("user","find the highest revenue")
    ]
})
```
