```python
from langchain_ollama import ChatOllama
from langchain_core.tools import tool
from langchain.agents import create_agent
import torch
```

```python
# !pip install "transformers[torch]" sentencepiece
```

```python
@tool
def summarize_text(text: str) -> str:
    """Summarize the given text."""
    from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

    model_name = "UX4567/Text-Summarizer-t5-small"

    tokenizer = AutoTokenizer.from_pretrained(model_name)
    summarizer = AutoModelForSeq2SeqLM.from_pretrained(model_name)

    
    inputs = tokenizer(
        text,  
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    outputs = summarizer.generate(
        **inputs,
        max_length=100,
        min_length=30,
        do_sample=False
    )

    summary = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return summary  
```

```python
tools = [summarize_text]
```

```python
model = ChatOllama(
    model = 'qwen2.5:0.5b',
    temperature=0.1
)
```

```python
agent = create_agent(
    model,
    tools,
    system_prompt="You are a text summarization agent. Use summarize_text whenever the user asks for a summary and return only the summary text."
)

```

```python
user_prompt='''summarize the given text :
        Artificial intelligence has transitioned from a theoretical concept into a cornerstone of modern global infrastructure.
        Early computing systems relied strictly on rigid, pre-programmed rules designed by human engineers to execute tasks.
        Today, modern machine learning architectures analyze massive datasets to discover intricate underlying patterns independently.
        This shift allows software to adapt dynamically to new inputs without requiring explicit manual reprogramming.
        Natural language processing enables computers to comprehend, interpret, and generate human languages with high precision.
        Computer vision systems now identify objects in real-time, powering autonomous vehicles and advanced medical diagnostics.
        Generative AI models can create realistic images, compose music, and write functional software code from text prompts.
        Industries like healthcare utilize these tools to accelerate drug discovery timelines and personalize patient treatment plans.
        Financial institutions deploy predictive algorithms to detect fraudulent transactions and optimize global market portfolios.
        Despite these massive technological breakthroughs, building reliable AI systems introduces significant technical hurdles.
        Training state-of-the-art foundation models demands immense computational power and massive data center infrastructure.
        Ensuring data privacy, mitigating algorithmic bias, and preventing hallucinations remain critical areas of ongoing research.
        Engineers are actively developing alignment techniques to ensure AI behaviors match human ethical standards.
        The future points toward autonomous multi-agent networks capable of collaborating to solve complex, open-ended tasks.
        Ultimately, the goal is to build safe systems that augment human intelligence and drive global scientific discovery.'''
```

```python
response = summary_agent(user_prompt)
```
