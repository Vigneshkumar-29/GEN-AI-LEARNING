```python
import os
import requests



API_URL = "https://router.huggingface.co/v1/chat/completions"
headers = {
    "Authorization": f"Bearer {os.environ['HF_TOKEN']}",
}

def query(payload):
    response = requests.post(API_URL, headers=headers, json=payload) # the post is used to give request and accept the response and update the value 
    return response.json()

# api url = path
# headers = authorization 
# json = the payload what the user want 

response = query({
    "messages": [
        {
            "role": "user", # role of the users
            "content": "What is the capital of France?" # user prompt 
        }
    ],
    "model": "deepseek-ai/DeepSeek-R1:novita"
})

#print(response["choices"][0]["message"])
print (response)
```
