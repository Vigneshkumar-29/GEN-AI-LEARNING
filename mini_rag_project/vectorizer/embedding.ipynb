```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
data = open('D:\Code\Gen-AI\Chunking\data.txt',encoding='utf-8').read()
```

```python
data
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 200,
    chunk_overlap = 40,
    # separators = ['\n\n','.','\n',''] # we can use any symbol like ! , [for larger doc we use this ]
)
chunk = text_splitter.split_text(data)
```

```python
vectoerizer = CountVectorizer() #it is used to covert the chunk into vector
# it is an ML preprocessing techniques 

vectors = vectoerizer.fit_transform(chunk)
```

```python
vectors  # it give in the form of sparse matrix we want to change into array 
```

```python
vectors.toarray() # we use this to convert to array 
```

```python
vectoerizer.get_feature_names_out() 
```
