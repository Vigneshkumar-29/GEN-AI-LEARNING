# fixed size chunk

```python
fixed_size_chunk = []
chunk_size = 50
for index in range(0,len(data),chunk_size):
    # print(data[index:index+chunk_size])
    fixed_size_chunk.append(data[index:index+chunk_size])
```

```python
print(fixed_size_chunk)
```

```python
len(fixed_size_chunk)
```

# chunk over lap

```python
#  chunk over lap size = 20 or 25% from the chunk size
# used in rag architecture 

fixed_size_chunk = []
chunk_size = 50
chunk_over_lap = 10
for index in range(0,len(data),chunk_size-chunk_over_lap):
    # print(data[index:index+chunk_size])
    fixed_size_chunk.append(data[index:index+chunk_size])
```

```python
print(fixed_size_chunk)
```
