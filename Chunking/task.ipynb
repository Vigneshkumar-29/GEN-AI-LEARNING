```python
import spacy
from nltk.stem import WordNetLemmatizer
```

```python
nlp = spacy.load("en_core_web_sm")
lemmatizer = WordNetLemmatizer()
```

```python
with open('data.txt','r') as file:
    text = file.read()
```

spacy tokenization

```python
doc = nlp(text)
```

spacy normalization

```python
spacy_tokens = [
    token.lemma_.lower()
    for token in doc
    if not token.is_stop
    # and (not token.is_punct or token.text == '.')
    # and not token.is_space
]
spacy_tokens=' '.join(spacy_tokens)
```

Chunking - over laping chunking

```python
chunk_size = 200
over_lap = 40
step = chunk_size-over_lap
chunk = []
for i in range(0,len(spacy_tokens),step):
    chunk.append(spacy_tokens[i:i+chunk_size])

chunk = ' '.join(chunk)
print(chunk)

```

```python
file = open('Task_clean.txt','w')
file.writelines(chunk)
file.close()
```
