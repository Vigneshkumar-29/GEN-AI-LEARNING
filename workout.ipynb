Reading the data.txt file 

```python
data = open('D:\Code\Gen-AI\data.txt',encoding='utf-8').read()
```

Convert the data into lower case 

```python
data = data.lower()
```

import the needed modules

```python
import re
import contractions
import string
import emoji
import textblob
import spacy
```

1. remove the extra space

```python
data = re.sub(r'\s{2,}','',data)
```

2. fixing the words by using the contraction i'm --> i am

```python
data = contractions.fix(data)
```

3. Removing the punctuation from the data except . dot

```python
str_punchuation = string.punctuation
str_punchuation = str_punchuation.replace('.','')
data = re.sub(f'[{str_punchuation}]','',data)
```

4. Removing the emoji from the data file 

```python
# data = emoji.demojize(data) # it covert the emoji into text

data = emoji.replace_emoji(data,'')  # it remove all the emoji in the data 
```

5. again remove some special character which are not removed in the punctuation stage

```python
data = re.sub(r'[^a-zA-Z0-9\s]','',data) # we can pass \. for dot if we want don't remove this.
```

Stop words

```python
from nltk.corpus import stopwords
```

```python
text = data.split()
```

```python
text
```

```python
stop_word = stopwords.words('english')
clean_text = []
for word in text:
    if word not in stop_word:
        clean_text.append(word)
data = ' '.join(clean_text)
```

```python
data
```

applying spacy

```python
nlp = spacy.load('en_core_web_sm') # This loads the spaCy English model. It doesn't tokenize your text yet.
```

```python
tokens = nlp(data)  # paCy processes data and the resulting Doc containing the tokens is stored in the variable tokens.
```

```python
tokens
```

chunking

```python
chunk_size = 200
over_lap = 40
step = chunk_size-over_lap
chunk=[]
for i in range(0,len(tokens),step):
    chunk.append(tokens[i:i+chunk_size].text)
chunk
```

we can use the list to find the no of chunk like which u want to see.

```python
print(chunk[1])
```

```python
with open('token_data.txt','w') as file:
    for c in chunk:
        file.write(c)
        file.write('\n\n')
```
